# Предобработка

```python
df_clean = df.copy()
df_clean['distance_km'] = df_clean['distance_km'].astype(str).str.replace(',', '.')
df_clean['order_sum'] = df_clean['order_sum'].astype(str).str.replace(',', '.')
df_clean['order_sum'] = df_clean['order_sum'].astype(str).str.replace('₽', '')
df_clean['order_sum'] = df_clean['order_sum'].astype(str).str.replace(' ', '')
df_clean = df_clean.drop_duplicates(subset=['order_id'], keep='first')
df_clean.loc[df_clean['prep_minutes'] < 0, 'prep_minutes'] = None
df_clean.loc[df_clean['delivery_minutes'] < 0, 'delivery_minutes'] = None
df_clean['distance_km'] = df_clean['distance_km'].astype('float64')
df_clean['order_sum'] = df_clean['order_sum'].astype('float64')
df_clean.loc[df_clean['distance_km'] < 0, 'distance_km'] = None
```

# Повторный анализ данных

```python
print(df_clean.shape)
display(df_clean.nunique())
print(df_clean.duplicated().sum())
print(df_clean.isna().sum())
display(df_clean.info())
print(df_clean.select_dtypes(include='number').columns)
```

Главный вопрос опоздания

```python
df_clean['is_late'] = df_clean['delivery_minutes'] > df_clean['promised_minutes']
d = df_clean[df_clean['status'] == 'delivered']

display(d.groupby('district')['is_late'].mean())
display(d.groupby('transport')['delivery_minutes'].agg(['mean','median']))
display(d.groupby('weather')['is_late'].mean())
display(d.groupby('courier_load')['is_late'].mean())
```
